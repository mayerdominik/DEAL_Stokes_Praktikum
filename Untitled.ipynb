```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize'] = [12, 8]
```

```python


data_rel = []
data_abs = []

sol_aprox = []
sol_true = []
with open("/home/f/Documents/DEAL_Stokes_Praktikum/rel_error_at_points.txt", "r") as file:
    for line in file.readlines():
        data_rel.append([float(x) for x in line.split(";")])

with open("/home/f/Documents/DEAL_Stokes_Praktikum/abs_error_at_points.txt", "r") as file:
    for line in file.readlines():
        data_abs.append([float(x) for x in line.split(";")])
        

with open("/home/f/Documents/DEAL_Stokes_Praktikum/aprox_sol_at_points.txt", "r") as file:
    for line in file.readlines():
        sol_aprox.append([float(x) for x in line.split(";")])
        

with open("/home/f/Documents/DEAL_Stokes_Praktikum/sol_at_points.txt", "r") as file:
    for line in file.readlines():
        sol_true.append([float(x) for x in line.split(";")])
        
data_rel = np.array(data_rel)
data_abs = np.array(data_abs)
sol_aprox = np.array(sol_aprox)
sol_true = np.array(sol_true)

data_rel_log = np.log(data_rel[:, 0])
data_rel[:, 0] = data_rel_log
```

```python
data = [data_rel, data_abs, sol_aprox, sol_true]
threshold = 6
error_set_indices = np.where(data_abs[:, 0] >= threshold)
```

```python
figure, axis = plt.subplots(2, int(len(data)/2)) 
titles = ["Log Rel Error", "Abs Error", f"Approx Solution, t = {threshold}", "True"]
for i, axs in enumerate(axis.flatten()):
    alpha = np.zeros(data[i][:, 0].shape) + 1
    if i == 2:
         alpha[error_set_indices] = 0.1
    plot = axs.scatter(data[i][:, 1], data[i][:, 2], c = data[i][:, 0], s = 10, alpha = alpha)
    figure.colorbar(plot, ax = axs)
    axs.title.set_text(titles[i])
figure.tight_layout()
```
